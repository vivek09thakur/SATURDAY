# Pros ⚡ and Cons ❗

Python have many pros over any other language. Lets what are some pros and cons of python.
## Pros ⚡


1. **Easy to use** : Its a compact and easy to use ***Object Oriented Programming*** language with easy syntax rules.

2. **Expressive Language** : Python easy syntax allow us to express a lot.

3. **Interpreted Language** : Its an interpreted language not a compile language.

4. **Varity of avaliable module and library** : Python has libraries and module for almost every work , which makes it my favourite language.

5. **Cross Platform** : Python programs can be run on any platfrom. No matter whatever its ***windows , linux or MacOS***. All you need a python interpreter.

6. **Open-Source** : Python is open source so we don't need to play to anyone to use it.

And last but not the least...

7. **Varity of uses** : It can be used in almost every field such as ***AI , ML , Data Science , Backend development , game development etc***.

---

## Cons ❗

1. **Not fast as others** : As the interpreter interpretes codes line by line so the execution time of program is larger than other languages such as ***C++ or Java***. Therefore python is prefered for prototyping and experiments or science.

2. **Not strong on type binding** : Python interpreter not so strong on catching `Type-mismatch` issues.

3. **Not easily convertible** : Python programs are ussually are more complex when it comes to convert into another language. Because of less syntax.

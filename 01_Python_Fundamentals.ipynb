## Tokens

In a passage or text every individual words and punctuation is known as tokens. Or If we say in programming , the smallest unit of a program is known as Tokens.

**Lets take some examples**

> - A simple sentence of english :
```
    Hello hii , how are you
```
Here , each and every word ( 'Hello','hii',',','how','are','you' ) is called tokens. Same goes for a python program :
> - A simple python program :

```python
    name =  'vivek'
    age = 18
    gf = 0.0000000
    print('his name was ' name,'he was ',age , 'with ',gf ,'girlfriend')
```
Here , every single word is known as token. In this code :
```python
    name =  'vivek' # here name is a variable and 'vivek' is a string literal
    age = 18 # age is also a variable of integer type and 18 is an integer literal
    gf = 0.0000000 # gf is a floating point or decimal variable with decimal or floating point literal
    print('his name was ' name,'he was ',age , 'with ',gf ,'girlfriend') # print is a keyword
```
Variables , Keywords , Literals , Indetifiers etc . They all are Tokens lets learn one by one.

**Python has following tokens :**
- Keywords
- Identifiers (names)
- Literals
- Operators
- Punctuators

### Identifiers
Identifiers are the fundamental building blocks of a computer program and they are generally used for naming conventions.

#### Indentifiers forming rules of python:
- An identifier is an long sequence of letters and digits.
- **The first character must be a letter , (_ , underscores are also allowed but not numbers).**
- Uppercase and lower-case letters are different.
- **The digit can be 0 - 9 but only after 1st character of identifier.**
- **It must be not a keyword**
- Must not have any special characters except `_`.
